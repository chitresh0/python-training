```python
# For loop
for letter in 'hello':
    print(letter)
```

```python
# range : printing number range
for num in range(1, 11):
    print(num)
```

```python
# if we provide single argument in range(), the range will start from 0.
for num in range(5):
    print(num)
```

```python
# range with gap: specific position
for num in range(1, 11, 4):
    print(num)
```

```python
# break and continue
# break: Exiting the loop
while True:
    name = input('Enter candidates name or type EXIT for exiting: ')
    if name == 'EXIT':
        break
    print('Hello ', name)

    
```

```python
# continue: skipping the process and continue the loop
for num in range(1, 20):
    if num % 5 == 0:
        continue
    print(num)
```
