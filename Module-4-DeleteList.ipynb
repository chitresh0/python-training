```python
# for deleting element from the list we use "del" keyword
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
del top_cities[3]
print(top_cities)
```

```python
# now if we want to keep first 3 elements in the list and delete the remaining elements
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
del top_cities[3:]
top_cities
```

```python
# if we want to delete all elements
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
del top_cities[:]
print(top_cities)
```

```python
# we can also delete the list with "del"
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
del top_cities
print(top_cities)
```
