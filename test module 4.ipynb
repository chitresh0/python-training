```python
sample_dict = {
    "mouth": "Mund",
    "finger": "Finger",
    "leg": "Bein",
    "hand": "Hand",
    "face": "Gesicht",
    "nose": "Nase"
}

while True:
    word = input('Enter a word in English or Exit: ')
    if word in sample_dict:
        print('Translation:', sample_dict[word])
    elif word == 'EXIT':
        print('Bye!')
        break
    else:
        print('No match!')
        
    
```
