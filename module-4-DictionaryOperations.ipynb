```python
# Dictionary Operations
```

```python
# definign a empty dictionary
grades = {}
```

```python
# adding grades of student in dictionary
grades['Ram'] = 'A'
```

```python
grades['Raj'] = 'B+'
```

```python
print(grades)
```

```python
# updating the dictionary value
grades['Raj'] = 'B'
```

```python
print(grades)
```

```python
# there is another opotion to update the dictionary 
# we can use update method to update the values
grades.update({'Raj': 'A'})
```

```python
print(grades)
```

```python
# we can also get the length of dictionary by using len function
len(grades)
```

```python
# we can delete the key by using del operator
del grades['Raj']
```

```python
print(grades)
```

```python
# Iterating the dictionary 
# there are multiple way to iterate the dictionary
grades = {}
grades['Ram'] = 'A'
grades['Raj'] = 'B'
```

```python
for ele in grades:
    print(ele)
```

```python
# so here we can see it only showing the keys of dictionary
```

```python
# we can also get the keys of dictionary by following way
for ele in grades.keys():
    print(ele)
```

```python
# we can also get the values of dictionary by following way
for ele in grades.values():
    print(ele)
```

```python
# get the access of both keys and values we can use the following method
for person, grade in grades.items():
    print(person, 'got', grade)
```

```python
for ele in grades.items():
    print(ele)
```

```python
# so in above example ele contians both key and value of each dictionary element
# we can get both key-value separate by indexing it
for ele in grades.items():
    print(ele[0], 'got', ele[1])
```
