```python
# Checking presence of element in list
userNames = ['raj', 'ram', 'sita', 'gita', 'hari']
yourName = input('Enter your name: ')
if yourName in userNames:
    print('you are a valid user')
else:
    print('your are not a valid user')
```

```python
# we can also check presence using "not in" key
userNames = ['raj', 'ram', 'sita', 'gita', 'hari']
yourName = input('Enter your name: ')
if yourName not in userNames:
    print('you are not a valid user')
else:
    print('your are a valid user')
```

```python
# Copying list
# example of coping variables
a = 'john'
b = a
a = 'mohan'
print(a, b)
```

```python
# in above example we have copied variable a value to vairable b 
# but after reasigning variable a it does not affect varaible b
# not let's see the same approach for a list
```

```python
# copying list
listA = [1, 2, 3]
listN = listA
listA[0] = -5
print('listA : ', listA, '\nlistN : ', listN)
```

```python
# here you can see after copying the listA value in listN when we changed the element of listA it automatically changed the element of listN too.
# this is becuase of when we copy the list it created a new variable but pointing to the same list

# for copying the list to another differnt list we can use slicing. 
```

```python
# copying list using slicing
listA = [1, 2, 3]
listN = listA[:]
listA[0] = -5
print('listA : ', listA, '\nlistN : ', listN)
```

```python
# here we have copied the listA into listN using slicing so both refer to different lists
```

```python
# we can also copy the a range of elements in list using slicing
listA = [1, 2, 3]
listN = listA[:2]
listA[0] = -5
print('listA : ', listA, '\nlistN : ', listN)
```
