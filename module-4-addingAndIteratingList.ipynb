```python
# to add elements in exising list we can use append() method.
```

```python
numberList = [1, 4, 5, 3, 9]
numberList.append(6)
print(numberList)
```

```python
# other than appending we can add new element to specific index 
numberList = [1, 4, 5, 3, 9]
numberList.insert(2, 8)
print(numberList)
```

```python
# Iterating list
# to read data from list we can iterate it by using loops
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
for city in top_cities:
    print('current city: ', city)
```

```python
# to get the index of the list we can iterate the range of list
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
for city_index in range(len(top_cities)):
    print('index : ', city_index, ' || current city: ', top_cities[city_index])
```

```python
# summation of list elements
spendings = [37, 32, 21, 40, 29, 34]
sum = 0
for spending in spendings:
    sum += spending
print('Money spent: ', sum)
```

```python
spendings = [1346.0, 987.50, 1734.40, 2567.0, 3271.45, 2500.0, 2130.0, 2510.30, 2987.34, 3120.50, 4069.78, 1000.0]
x, y, z = 0, 0, 0
for spending in spendings:
    if spending < 1000.0:
        x += 1
    elif spending > 2500.0:
        z += 1
    else:
        y += 1

print('Numbers of months with low spendings: ' + str(x) + ', normal spendings: ' + str(y) + ', high spendings: ' + str(z) + '.')
```
