```python
# Swapping elements fo list

# normal swapping 
a = 10
b = 20
temp = a
a = b
b = temp
print(a, b)
```

```python
# in python there is not required an extra variable we can swap the variables like this
a = 10
b = 20
a, b = b, a
print(a, b)
```

```python
# swapping list elements or inter-changing positions
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
top_cities[0], top_cities[4] = top_cities[4], top_cities[0]
print(top_cities)
```

```python
# we can sort the list
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
top_cities.sort()
print(top_cities)
```

```python
# we can sort the numbers form lowest to greatest
numbers = [4, 2, 8, 0, -5, 6]
numbers.sort()
print(numbers)
```

```python
# we can reverse the sorting form greatest to lowest
numbers = [4, 2, 8, 0, -5, 6]
numbers.sort(reverse=True)
print(numbers)
```

```python
# sorted() : return a new sorted list keep the original unchanged
# list.sort() : sorts the original list
top_cities = ['Hyderabad', 'Delhi', 'Mumbai', 'Pune', 'Chennai']
print(sorted(top_cities))
print(top_cities)
```
