```python
cells = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
                            
```

```python
cells[0]
```

```python
cells[1]
```

```python
cells[0][0]
```

```python
cells = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for x in cells:
    print('Element : ', x)
```

```python
cells = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for x in cells:
    for y in x:
        print('Element: ', y)
```

```python
table = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for row in table:
    for cell in row:
        print(cell,  end='\t')
    print()
```

```python
# creating nested list using comprehensions
table = [[i for i in range(1, 6)] for j in range(4)]
print(table)
```
