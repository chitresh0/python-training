```python
# Keyword arguments / Named arguments
```

```python
print('hello')
print('python')
```

```python
# using keyword argument => end=
print('hello', end=', ')
print('python')
```

```python
## a keyward argument is optional argument and its default value is always a new line character (i.e. '\n')
```

```python
## there is onemore keyward argument in print function is "sep" which specify the separator between the value printed in the output
## the default value for sep argument is space character
```

```python
print('my name is', 'john')
```

```python
# here automatically a space is added between both the argument
```

```python
# now we will add user define sep keyward argument
print('my name is', 'john', sep='-')
```

```python
# we can use both sep and end argument at same time 
print('my name is', 'john', sep='-', end='!')
```
