```python
# List comprehensions
# to create a list which contains numbers 1 to 100 
numbers = []
for i in range(1, 101):
    numbers.append(i)

print(numbers)
```

```python
# so here we created a empty list and then assigned values to it

# but this is very lenghty way
# insted of above approach we can use comprehensions in python
```

```python
numbers = [i for i in range(1, 101)]
print(numbers)
```

```python
# now we will generate a list which contains numbers 1 to 100 but exclude numbers which is devided by 3
numbers = [i for i in range(1, 101) if i % 3 != 0]
print(numbers)
```
