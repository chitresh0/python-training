```python
state1 = ('Telangana', 'Hyderabad', 4.4)
```

```python
state2 = ('Maharastra', 'Mumbai', 2.3)
```

```python
state3 = ('Punjab', 'Chandigarh', 1.5)
```

```python
# List of tuples
states = [state1, state2, state3]
```

```python
for state in states:
    print('Name:', state[0], ', capital:', state[1], ', population:', state[2])
```

```python
# list in tuple
user_data = ('Ram', 'India', 1994, [87.1, 79.3, 88.1])
```

```python
print(user_data)
```

```python
user_data[3].append(83.2)
```

```python
print(user_data)
```

```python
# so here we can see we can not modify the Tuple but we can modify a list inside a tuple
# Tuples are immutable but list inside a tuple is mutable.
```
