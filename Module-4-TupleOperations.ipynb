```python
# getting length of tuple
user_data = ('Ram', 'India', 1993)
print(len(user_data))
```

```python
 # we can use in and not in operator same as list
user_data = ('Ram', 'India', 1993)
if 'India' in user_data:
    print('This person is from India')
```

```python
# we can also use for loop to iterate data
user_data = ('Ram', 'India', 1993)
for element in user_data:
    print(element)
```

```python
# same like string and list we can add and multiply the tuple elements
user_data = ('Ram', 'India', 1993) + ('teacher', 'male')
print(user_data)
```

```python
# we can not append the new elements like list but we can add 
# multiply
number = (0, 1) * 10
print(number)
```

```python
# list can be use for similar data types but tuples can be use for multiple data type
user_data = ('Ram', 'India', 1993, 5.20)
print(user_data)
```
