```python
# difference between list and tupple 
# list - mutable            but in            tupple - immutable
# list - add and modify the elements            but in            tuple - can not add and modify the elements
```

```python
new_tuple = ('Raj', 'India', 1993)
```

```python
new_tuple.append('student')
```

```python
# even we can not delete the elements from tuple
del new_tuple[0]
```

```python
# we can read the data from tuple similar to list
print(new_tuple[0])
```

```python
# but we can to modify the element
new_tuple[0] = 'Ram'
```

```python
# we can assign the new value to a tuple
new_tuple = ('Raj', 'India', 1993)
new_tuple = ('Ram', 'India', 1983)
print(new_tuple)
```
