```python
# Dictionary example
email = {
    'ram': 'ram@ex.com',
    'raj': 'raj@ex.com',
    'geeta': 'geeta@ex.com',
    'sita': 'sita@ex.com'
}
```

```python
print(email)
```

```python
print(email['raj'])
```

```python
# in dictionary keys are unique if we provide one key with multiple value then it overwrite the key with last value
country_code = {
    'India': '001',
    'Amaerica': '002',
    'Australia': '003',
    'Africa': '004',
    'Africa': '005'
}
```

```python
print(country_code)
```

```python
# keys can be intiger, float and tuples but not list because list are mutable
# we can use list as value of the keys
```

```python
ranking = {
    1: 'India',
    2: 'America',
    3: 'Russia',
    4: 'China'
}
```

```python
print(ranking[3])
```

```python
ranking = {
    [1]: 'India',
    [2]: 'America',
    [3]: 'Russia',
    [4]: 'China'
}
```

```python
# you can see in above example a list can't be a key for a dictionary 
# but we can use list as a value 
ranking = {
    'India': [5, 3, 4, 4],
    'America': [5, 4, 5, 4],
    'Russia': [5, 4, 3, 4],
    'China': [5, 5, 4, 4]
}
```

```python
print(ranking['China'])
```
