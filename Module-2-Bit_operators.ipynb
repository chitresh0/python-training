```python
# there are 6 bitwise operators in python
# & ampersand - logical AND gate
# | bar - logical OR gate
# ~ tild - logical NOT gate
# ^ caret - logical XOR gate
# << left shift 
# >> ritht shift
```

```python
print (0 & 0)
print (0 & 1)
print (1 & 0)
print (1 & 1)
```

```python
print (0 | 0)
print (0 | 1)
print (1 | 0)
print (1 | 1)
```

```python
print (0 ^ 0)
print (0 ^ 1)
print (1 ^ 0)
print (1 ^ 1)
```

```python
# ~ is an unary operator
# ~x = -x - 1
print(~2)
```

```python
# x << y => left shift the bit by y position
# for example if we have a number 12, so in binary it look like => 1100  
# and if we do left shift this number by 1 i.e. 12 << 1, means we are shifting the bits left side for 1 position 
# so in binary it will 1100 << 1 => 11000 which is equivalent to 24 in decimal

# so in decimal we can say x << y means x * y^2 
12 << 1
```

```python
12 << 2
```

```python
12 << 4
```

```python
# Right shift operator >>
# x >> y => Right shift the bit by y position
# so 12 >> 1 means shifting the bits right side for 1 position
# in binary it will be 1100 >> 1 => 0110 which is equivalent to 6
# so in decimal we can assume x >> y means x / y^2 (devide the number by cetain power of 2)
12 >>1
```

```python
12 >> 2
```

```python
# these shifting logic in decimal is only for quick calculation it is not the currect way to calculate
# because these binary operators only works with binary numners
# so in certain scenario these decimal logics may fail 
# for example if we try 12 >> 3 so according our decimal logic it should be 12 / 2^3 => 12 / 8 => 1.5 
# so we are getting the answer in floating which is not currect 
# according our binary logic it should work like => 1100 >> 3 => 0001 which is equivalet to 1 in decimal
12 >> 3
```

```python
# and if we take 12 << 4 => 1100 << 4 => 0000 => equivalent to 0 in decimal
12 >> 4 
```
